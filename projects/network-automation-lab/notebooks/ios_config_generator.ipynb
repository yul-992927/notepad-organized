```python
# ============================================
# Cisco IOS 설정 명령어 자동 생성기
# GNS3 / Jupyter Notebook 용
# ============================================


def span():
    session = input("SPAN Session 번호 [예: 1]: ")
    source = input("Source Interface [예: fa0/1]: ")
    destination = input("Destination Interface [예: fa0/10]: ")
    direction = input("방향 (rx / tx / both) [기본 both]: ").strip().lower()

    if direction == "":
        direction = "both"

    return [
        "conf t",
        f"monitor session {session} source interface {source} {direction}",
        f"monitor session {session} destination interface {destination}"
    ]


def port_security():
    interface = input("적용 Interface [예: fa0/1]: ")
    maximum = input("허용 MAC 최대 개수 [예: 1]: ")
    violation = input(
        "Violation 모드 (protect / restrict / shutdown) [예: restrict]: "
    ).strip().lower()

    return [
        "conf t",
        f"interface {interface}",
        "switchport mode access",
        "switchport port-security",
        f"switchport port-security maximum {maximum}",
        "switchport port-security mac-address sticky",
        f"switchport port-security violation {violation}",
        "no shutdown"
    ]


def ppp_pap():
    interface = input("PPP Interface [예: s0/0]: ")
    local_name = input("현재 Router hostname [예: R1]: ")
    remote_name = input("상대 Router hostname [예: R2]: ")
    password = input("PAP Password: ")

    return [
        "conf t",
        f"username {remote_name} password {password}",
        f"interface {interface}",
        "encapsulation ppp",
        "ppp authentication pap",
        f"ppp pap sent-username {local_name} password {password}",
        "no shutdown"
    ]


def ppp_chap():
    interface = input("PPP Interface [예: s0/0]: ")
    remote_name = input("상대 Router hostname [예: R2]: ")
    password = input("CHAP Password: ")

    return [
        "conf t",
        f"username {remote_name} password {password}",
        f"interface {interface}",
        "encapsulation ppp",
        "ppp authentication chap",
        "no shutdown"
    ]


def rip():
    commands = [
        "conf t",
        "router rip",
        "version 2",
        "no auto-summary"
    ]

    count = int(input("RIP에 등록할 Network 개수: "))

    for i in range(count):
        network = input(
            f"Network {i + 1} [예: 198.18.0.0]: "
        )
        commands.append(f"network {network}")

    return commands


def eigrp():
    as_num = input("EIGRP AS 번호 [예: 100]: ")

    commands = [
        "conf t",
        f"router eigrp {as_num}"
    ]

    count = int(input("EIGRP에 등록할 Network 개수: "))

    for i in range(count):
        network = input(
            f"Network {i + 1} [예: 198.18.10.0]: "
        )

        wildcard = input(
            f"Wildcard Mask {i + 1} [예: 0.0.0.3]: "
        )

        commands.append(
            f"network {network} {wildcard}"
        )

    commands.append("no auto-summary")

    return commands


def eigrp_active_timer():
    as_num = input("EIGRP AS 번호 [예: 100]: ")
    timer = input("Active Timer (분) [예: 5]: ")

    return [
        "conf t",
        f"router eigrp {as_num}",
        f"timers active-time {timer}"
    ]


def eigrp_stub():
    as_num = input("EIGRP AS 번호 [예: 100]: ")

    print("""
===== EIGRP Stub 옵션 =====

1. connected summary
2. connected
3. connected summary static
4. receive-only
""")

    stub_choice = input("Stub 옵션 선택: ")

    stub_options = {
        "1": "connected summary",
        "2": "connected",
        "3": "connected summary static",
        "4": "receive-only"
    }

    stub_option = stub_options.get(stub_choice)

    if stub_option is None:
        print("잘못된 Stub 옵션입니다.")
        return []

    return [
        "conf t",
        f"router eigrp {as_num}",
        f"eigrp stub {stub_option}"
    ]


def eigrp_variance():
    as_num = input("EIGRP AS 번호 [예: 100]: ")
    variance = input("Variance 값 [예: 2]: ")
    paths = input("Maximum Paths [예: 4]: ")

    return [
        "conf t",
        f"router eigrp {as_num}",
        f"variance {variance}",
        f"maximum-paths {paths}"
    ]


def passive_interface():
    print("""
===== Passive Interface Protocol =====

1. EIGRP
2. RIP
""")

    protocol = input("Protocol 선택: ")
    interface = input("Passive Interface [예: fa0/0]: ")

    if protocol == "1":

        as_num = input("EIGRP AS 번호 [예: 100]: ")

        return [
            "conf t",
            f"router eigrp {as_num}",
            f"passive-interface {interface}"
        ]

    elif protocol == "2":

        return [
            "conf t",
            "router rip",
            f"passive-interface {interface}"
        ]

    else:
        print("잘못된 Protocol 선택입니다.")
        return []


def default_route():
    next_hop = input(
        "Default Route Next-Hop IP [예: 198.18.10.2]: "
    )

    floating = input(
        "Floating Static으로 설정? (y/n) [기본 n]: "
    ).strip().lower()

    if floating == "y":

        ad = input(
            "Administrative Distance [예: 200]: "
        )

        return [
            "conf t",
            f"ip route 0.0.0.0 0.0.0.0 {next_hop} {ad}"
        ]

    else:

        return [
            "conf t",
            f"ip route 0.0.0.0 0.0.0.0 {next_hop}"
        ]


# ============================================
# 메인 메뉴
# ============================================

print("""
=================================
       Cisco 설정 메뉴
=================================

1. SPAN
2. Port Security
3. PPP PAP
4. PPP CHAP
5. RIP
6. EIGRP
7. EIGRP Active Timer
8. EIGRP Stub
9. EIGRP Variance
10. Passive Interface
11. Default Route

=================================
""")

choice = input("설정할 기술 번호 입력: ").strip()


# ============================================
# 메뉴 선택
# ============================================

if choice == "1":
    commands = span()

elif choice == "2":
    commands = port_security()

elif choice == "3":
    commands = ppp_pap()

elif choice == "4":
    commands = ppp_chap()

elif choice == "5":
    commands = rip()

elif choice == "6":
    commands = eigrp()

elif choice == "7":
    commands = eigrp_active_timer()

elif choice == "8":
    commands = eigrp_stub()

elif choice == "9":
    commands = eigrp_variance()

elif choice == "10":
    commands = passive_interface()

elif choice == "11":
    commands = default_route()

else:
    print("잘못된 메뉴 번호입니다.")
    commands = []


# ============================================
# 생성된 Cisco 명령어 출력
# ============================================

if commands:

    print("\n=================================")
    print("      생성된 Cisco 명령어")
    print("=================================\n")

    for cmd in commands:
        print(cmd)
```
